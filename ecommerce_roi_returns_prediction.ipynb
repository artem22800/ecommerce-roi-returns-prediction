```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
marketing_spends = pd.read_csv('marketing_spend.csv')
orders = pd.read_csv('orders.csv')
```

```python
marketing_spends
```

```python
orders
```

```python
orders['order_date'] = pd.to_datetime(orders['order_date'])
marketing_spends['month'] = pd.to_datetime(marketing_spends['month'], format = '%Y-%m')
```

```python
month_oreder_velue = orders.groupby(['marketing_channel', orders['order_date'].dt.to_period('M')]).final_order_value.sum().reset_index()
ROI = (month_oreder_velue.final_order_value - marketing_spends.marketing_spend).reset_index()
```

```python
ROI['channel'] = marketing_spends['channel']
ROI['month'] = marketing_spends['month']
ROI.drop(columns = 'index', inplace = True)
ROI['ROI_percentage'] = (month_oreder_velue.final_order_value - marketing_spends.marketing_spend) / month_oreder_velue.final_order_value * 100
```

```python
ROI.groupby('channel').ROI_percentage.mean()
```

самыее выгодные каналы по относительному ROI это email и referral, ихние ROI более чем 90%. убыточных каналов нет.
